**List of dictionaries, where each dictionary represents a student.**

```python
students = [{'name': 'Neto', 'score': 95, 'is_present': True},

    {'name': 'Netty', 'score': 76, 'is_present': True},

    {'name': 'Miracle','score': 0, 'is_present': False}]
```

Loop through your list of students. For each student:
    *   Print the student's name. **bold text**

```python
for student in students:
   print(f"Student Name: {student['name']}")
   if student['is_present'] == False:
       print('Student was absent.')
   else:
       score = student['score']
       if score >= 90:
           print('Excellent! Grade A.')
       elif score >= 80:
           print('Very Good! Grade B.')
       elif score >= 70:
           print('Good! Grade C.')
       elif score >= 60:
           print('Pass. Grade D.')
       else:
           print('Fail. Grade F. Needs to re-sit.')
```
