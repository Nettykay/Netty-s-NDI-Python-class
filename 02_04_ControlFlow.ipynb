### If / Else statements

```python
# Grading sheet
score = 35

if score >= 70:
  print('Grade A')
elif score >= 60:
  print('Grade B')
elif score >= 50:
  print('Grade C')
elif score >= 45:
  print('Grade D')
elif score >=40:
  print('Grade E')
else:
  print('Grade F')
```

Here, score is the variable. ensure your elif statement begings from the end.

```python
a = True
if a:
    print('It is true!')
    print('Also print this')
else:
    print('It is false!')
print('Always print this')
```

```python
# Eligibility check
age = 20
has_id = True

if age >= 18 and has_id == True:
    print("You are eligible to vote")

elif age < 18:
    print("You are too young to vote")

elif age >= 18 and has_id == False:
    print("Not eligible: ID required")
```

### For loops

```python
a = [1,2,3,4,5]
for number in a:
    print(number)
```

Here, a is the variable and the numbers in the squared bracket are the indexes. The code above means for each individual number in a, print number. If print number is changed to print a, it prints everything in a which will be [1,2,3,4,5] 5 times but this one above will print the individual numbers.

```python
a=[1,2,3,4,5]
for number in a:
  print(number + 3)
```

### While loops

```python
a = 0
while a < 5:
    print(a)
    a = a + 1
```

Provided that a is < 5, print the a, then a = a + 1

0 + 1 = 1
1 + 1 = 2

it keeps goint on till the addition becomes 5 or greater then the code stops running.

In while loop, ensure to write a condition that can definitely turn to false.
