```python
import math
```

## Functions

```python
def performOperation(num1, num2, operation):
    if operation == 'sum':
        return num1 + num2
    if operation == 'multiply':
        return num1 * num2

performOperation(2, 3, 'sum')
```

### Named Parameters

```python
def performOperation(num1, num2, operation='sum'):
    if operation == 'sum':
        return num1 + num2
    if operation == 'multiply':
        return num1 * num2

performOperation(2, 3)
```

```python
def performOperation(num1, num2, operation='sum', message='Default message'):
    print(message)
    if operation == 'sum':
        return num1 + num2
    if operation == 'multiply':
        return num1 * num2

performOperation(2, 3, message='A new message!', operation='multiply')
```

### *args

```python
def performOperation(*args):
    print(args)

performOperation(1,2,3)
```

*args means accepting multiple arguments. When * is seen behind a variable name, it accepts as many arguments as possible.

```python
def performOperation(*Neto):
  print(Neto)

performOperation(1,2,3)
```

For keyword arguments, we use **kwargs

```python
performOperation(1,2,3, operation='sum')
```

### **kwargs

```python
def performOperation(*args, **kwargs):
    print(args)
    print(kwargs)
performOperation(1,2,3, operation='sum')
```

```python
def performOperation(*args, operation='sum'):
    if operation == 'sum':
        return sum(args)
    if operation == 'multiply':
        return math.prod(args)

performOperation(1,2,3,6, 7, 8, operation='sum')
```
