## Classes

```python
class Dog:
    def __init__(self, name):
        self.name = name
        self.legs = 4

    def speak(self):
        print(self.name + ' says: Bark!')


```

A class houses fxns that are related.
Here, Dog is the class and we want to create a class that will house name. self is a necessary parameter in class and the init method is necessary as it houses necessary info that the class must have. The name as a parameter becomes a variable.
self.name = name
(name becomes an attribute).

```python
my_dog = Dog('Rover')
another_dog = Dog('Fluffy')
```

```python
my_dog.speak()
```

```python
another_dog.speak()
```
