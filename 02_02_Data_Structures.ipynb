## Dictionaries

```python
my_dictionary = {
    'apple': 'A red fruit',
    'bear': 'A scary animal'
}
```

my_dictionary is the variable. The index here is the key(apple) and the value is the item stored in the key. So, the key is used to access the value. To access the value, you call up the key.

```python
my_dictionary['bear']
```

```python
my_dictionary = {
    'apple': 'A red fruit',
    'bear': 'A scary animal',
    'apple': 'Sometimes a green fruit'
}
```

```python
my_dictionary['apple']
```

```python
my_dictionary = {'apple': 'A sweet fruit', 'Neto': 'A beautiful lady'}
```
